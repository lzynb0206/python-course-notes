## 1. 从普通列表到 NumPy 数组

Python 列表适合保存通用对象；NumPy 的 `ndarray` 更适合保存同一种类型的大量数字。数组支持向量化运算：一个表达式就能作用到所有元素，不必手写循环。


```python
import numpy as np
import pandas as pd

temperatures = [18.5, 20.0, 21.5, 23.0]
array = np.array(temperatures)

array
```

## 2. 向量化运算

下面把摄氏温度整体转换为华氏温度。`array * 9 / 5 + 32` 会逐元素计算，代码直接表达了公式本身。


```python
fahrenheit = array * 9 / 5 + 32
fahrenheit
```

## 3. Pandas 给数据加上标签

NumPy 更关注数值和形状；Pandas 在此基础上提供行索引、列名以及表格操作。`Series` 像带索引的一列数据，`DataFrame` 像一张二维表。


```python
weather = pd.DataFrame(
    {
        "城市": ["北京", "上海", "广州"],
        "温度": [18.5, 22.0, 27.5],
        "是否下雨": [False, True, False],
    }
)
weather
```

## 4. 用列名完成分析

选取一列后得到 `Series`，可以继续做平均值、筛选等分析。表达式 `weather["温度"] >= 22` 会得到一组布尔值，再用它筛选整张表。


```python
hot_cities = weather.loc[weather["温度"] >= 22, ["城市", "温度"]]
hot_cities
```

## 三者如何选择

| 工具 | 适合场景 | 主要特点 |
|---|---|---|
| Python 列表 | 少量通用数据 | 可混合类型，写法灵活 |
| NumPy 数组 | 大量数值计算 | 同类型、速度快、支持多维数组 |
| Pandas | 表格清洗与分析 | 有行列标签，擅长缺失值和关联操作 |


```python
comparison = pd.DataFrame(
    {
        "工具": ["list", "ndarray", "DataFrame"],
        "有标签": [False, False, True],
        "适合数值批量计算": [False, True, True],
    }
)
comparison
```

## 本节小结

NumPy 负责高效的数组计算，Pandas 负责带标签的表格分析。后面的 NumPy 知识会成为理解 Pandas 的基础。
