## 1. `np.array()` 与 `dtype`

`np.array()` 会尝试推断统一的数据类型。指定 `dtype` 可以控制整数位数或浮点精度。


```python
import numpy as np

integers = np.array([1, 2, 3], dtype=np.int32)
decimals = np.array([1, 2, 3], dtype=np.float64)

integers, integers.dtype, decimals, decimals.dtype
```

## 2. 按固定内容创建

`zeros`、`ones`、`full` 和 `eye` 常用于初始化。`empty` 只分配内存，不保证初始值，因此不能把它当成全零数组。


```python
zeros = np.zeros((2, 3))
ones = np.ones((2, 3), dtype=int)
filled = np.full((2, 3), 7)
identity = np.eye(3)

zeros, ones, filled, identity
```

## 3. `arange()` 与 `linspace()`

`arange(start, stop, step)` 更像 `range()`，关注步长；`linspace(start, stop, num)` 关注要生成多少个等距点，而且默认包含终点。


```python
by_step = np.arange(0, 10, 2)
by_count = np.linspace(0, 1, 5)

by_step, by_count
```

## 4. 创建有规律的二维数组

实际练习里经常先生成连续数字，再通过 `reshape()` 得到所需形状。


```python
grid = np.arange(1, 13).reshape(3, 4)
grid
```

## 5. 可复现的随机数

推荐使用 `default_rng()` 创建随机数生成器。指定种子后，每次运行都会得到相同结果，便于学习、测试和复现实验。


```python
rng = np.random.default_rng(42)
random_integers = rng.integers(1, 10, size=(2, 4))
random_normal = rng.normal(loc=0, scale=1, size=5)

random_integers, random_normal
```

## 6. 类型转换

`astype()` 返回转换后的新数组。浮点数转整数时会直接截掉小数部分，不是四舍五入。


```python
values = np.array([1.2, 2.8, -3.6])
truncated = values.astype(int)
rounded = np.rint(values).astype(int)

truncated, rounded
```

## 本节小结

根据问题选择创建方式：固定值用 `zeros/ones/full`，等差序列用 `arange/linspace`，实验数据用带种子的随机数生成器。
