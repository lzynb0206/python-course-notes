## 1. Series折线图

先用固定数据演示：


````python
data = pd.Series(
    [1, -2, 3, 2, -1],
    index=[0, 1, 2, 3, 4]
)

print(data)
````

输出：

```text
0    1
1   -2
2    3
3    2
4   -1
dtype: int64
```

累加：


````python
data = data.cumsum()

print(data)
````

输出：

```text
0    1
1   -1
2    2
3    4
4    3
dtype: int64
```

计算过程：

```text
第0个：1
第1个：1 + (-2) = -1
第2个：1 + (-2) + 3 = 2
第3个：1 + (-2) + 3 + 2 = 4
第4个：1 + (-2) + 3 + 2 + (-1) = 3
```

画图：


````python
data.plot()
plt.show()
````

图中：

```text
横轴x → Series索引0、1、2、3、4
纵轴y → 1、-1、2、4、3
```

连线关系：

```text
(0,1) → (1,-1) → (2,2) → (3,4) → (4,3)
```

## 2. 原视频的随机折线图


````python
data = pd.Series(
    np.random.randn(1000),
    index=np.arange(1000)
)

data = data.cumsum()

data.plot()
plt.show()
````

解释：

```python
np.random.randn(1000)
```

生成1000个服从标准正态分布的随机数。

```python
data.cumsum()
```

将数据累计，图形看起来像随机上下运动的轨迹。

一定要注意：

```python
data.cumsum()
```

只计算并返回结果，不会默认修改 `data`。

所以应该写：


````python
data = data.cumsum()
````

或者：

```python
data.cumsum().plot()
```

## 3. DataFrame画多条折线


````python
data = pd.DataFrame({
    "A": [1, 3, 2, 5],
    "B": [4, 2, 5, 3],
    "C": [2, 4, 1, 6]
})

print(data)
````

输出：

```text

0  1  4  2

2  2  5  1
3  5  3  6
```

画图：


````python
data.plot()
plt.show()
````

会画出三条线：

```text
A列 → 一条线
B列 → 一条线
C列 → 一条线
```

横轴默认使用DataFrame索引：

```text
0、1、2、3
```

````python
data.plot(
    title="成绩变化",
    marker="o"
)

plt.xlabel("考试次数")
plt.ylabel("成绩")
plt.show()
````

## 5. 散点图


````python
data = pd.DataFrame({
    "A": [1, 2, 3, 4],
    "B": [2, 4, 5, 8],
    "C": [8, 6, 4, 2]
})

ax = data.plot.scatter(
    x="A",
    y="B",
    color="DarkBlue",
    label="Class1"
)

data.plot.scatter(
    x="A",
    y="C",
    color="LightGreen",
    label="Class2",
    ax=ax
)

plt.show()
````

第一组点：

```text
A作为x
B作为y

(1,2)、(2,4)、(3,5)、(4,8)
```

第二组点：

```text
A作为x
C作为y

(1,8)、(2,6)、(3,4)、(4,2)
```

关键参数：

```python
ax=ax
```

表示把第二组散点画在第一张图的坐标轴上。

如果不传 `ax=ax`，通常会新建另一张图。

## 6. 柱状图


````python
scores = pd.Series(
    [90, 85, 92],
    index=["小明", "小红", "小刚"]
)

scores.plot(
    kind="bar",
    title="学生成绩"
)

plt.ylabel("分数")
plt.show()
````

会显示3根柱子。

横轴：

```text
小明、小红、小刚
```

纵轴：

```text
90、85、92
```

横向柱状图：


````python
scores.plot(kind="barh")
plt.show()
````

## 7. 直方图


````python
data = pd.Series([
    60, 65, 70, 70, 72,
    75, 80, 80, 85, 90
])

data.plot(
    kind="hist",
    bins=5
)

plt.show()
````

直方图用于观察数值分布。

```python
bins=5
```

表示大致分成5个区间统计数量。

注意：

```text
柱状图bar → 比较不同类别
直方图hist → 查看连续数值的分布
```

# `concat`和`merge`最终对比

这是最后几集最容易混淆的地方。

## concat

```python
pd.concat([df1, df2], axis=0)
```

主要是直接拼接：

```text
df1
df2
 ↓
上下堆叠
```

或者：

```python
pd.concat([df1, df2], axis=1)
```

左右堆叠。

---

## merge

```python
pd.merge(df1, df2, on="id")
```

主要是根据某个键寻找对应关系：

```text
df1中id=2的数据
        +
df2中id=2的数据
        ↓
合成一行
```

记忆：

```text
concat → 拼积木
merge  → 查身份证配对
```

| 左右拼接 | `pd.concat(..., axis=1)` |
| 根据键连接 | `pd.merge(..., on="id")` |
| 折线图 | `df.plot()` |
| 散点图 | `df.plot.scatter()` |
| 显示图片 | `plt.show()` |
| 保存图片 | `plt.savefig()` |

视频中三处必须更新的旧写法：

```python
# 旧：已经删除
df.ix[...]

# 新
df.loc[...]
df.iloc[...]
```

```python
# 旧：链式赋值可能出问题
df.B[df.A > 4] = 0

# 新
df.loc[df["A"] > 4, "B"] = 0
```

```python
# 旧：DataFrame.append已经删除
df1.append(df2)

# 新
pd.concat([df1, df2])
```
