## 1. 准备两个数组


````python
import numpy as np

a = np.array([10, 20, 30, 40])
b = np.arange(4)

print(a)
print(b)
````

输出：

```text
[10 20 30 40]
[0 1 2 3]
```

它们的形状都是：


````python
print(a.shape)
print(b.shape)
````

```text

(4,)
```

## 4. 对应位置相乘


````python
c = a * b

print(c)
````

## 7. 数组和一个数字运算

## 8. 数学函数


````python
b = np.array([0, 1, 2, 3])

print(np.sin(b))
````

输出近似为：

```text
[0.         0.84147098 0.90929743 0.14112001]
```

注意：

```python
np.sin()
```

使用的是弧度，不是角度。

如果想计算30度的正弦值：


````python
angle = 30

result = np.sin(np.deg2rad(angle))

print(result)
````

输出约为：

```text
0.49999999999999994
```

理论值就是：

```text
0.5
```

出现很长的小数是浮点数精度现象。

可以四舍五入：


````python
print(np.round(result, 2))
````

输出：

```text
0.5
```

## 9. 数组进行条件判断


````python
b = np.arange(4)

print(b)
print(b < 3)
````

输出：

```text
[0 1 2 3]
[ True  True  True False]
```

NumPy分别判断每一个元素：

```text
0 < 3 → True
1 < 3 → True
2 < 3 → True
3 < 3 → False
```

得到的是布尔数组。

查看类型：


````python
result = b < 3

print(result.dtype)
````

输出：

```text
bool
```

## 10. 使用布尔数组筛选数据


````python
b = np.array([0, 1, 2, 3, 4, 5])

condition = b < 3

print(condition)
print(b[condition])
````

输出：

```text
[ True  True  True False False False]
[0 1 2]
```

通常直接写：


````python
print(b[b < 3])
````

输出：

```text
[0 1 2]
```

执行过程：

```text
b       = [0,    1,    2,     3,     4,     5]
b < 3   = [True, True, True, False, False, False]
保留结果 = [0,    1,    2]
```

这叫布尔索引，后面处理数据时非常重要。

## 11. `*`和矩阵乘法的区别

准备两个二维数组：


````python
a = np.array([
    [1, 1],
    [0, 1]
])

b = np.array([
    [0, 1],
    [2, 3]
])

print(a)
print(b)
````

输出：

```text
[[1 1]
 [0 1]]

[[0 1]
 [2 3]]
```

### 对应位置相乘


````python
print(a * b)
````

输出：

```text
[[0 1]
 [0 3]]
```

计算方式：

```text
1×0  1×1
0×2  1×3
```

所以得到：

```text
0 1
0 3
```

---

### 标准矩阵乘法

原视频使用：


````python
print(np.dot(a, b))
````

输出：

```text
[[2 4]
 [2 3]]
```

也可以写：


````python
print(a.dot(b))
````

输出相同：

```text
[[2 4]
 [2 3]]
```

现代代码中二维矩阵乘法还常写：


````python
print(a @ b)
````

输出：

```text
[[2 4]
 [2 3]]
```

第一行第一列的计算：

```text
1×0 + 1×2 = 2
```

第一行第二列：

```text
1×1 + 1×3 = 4
```

第二行第一列：

```text
0×0 + 1×2 = 2
```

第二行第二列：

```text
0×1 + 1×3 = 3
```

所以结果为：

```text
[[2 4]
 [2 3]]
```

务必记住：

```text
a * b       → 对应位置相乘
a @ b       → 矩阵乘法
np.dot(a,b) → 二维情况下也是矩阵乘法
```

## 12. 求和、最大值和最小值

为了让输出固定，先不用随机数：


````python
a = np.array([
    [1, 2, 3, 4],
    [5, 6, 7, 8]
])

print(a)
````

输出：

```text
[[1 2 3 4]
 [5 6 7 8]]
```

全部元素求和：


````python
print(np.sum(a))
````

输出：

```text
36
```

计算：

```text
1+2+3+4+5+6+7+8 = 36
```

最小值：


````python
print(np.min(a))
````

输出：

```text
1
```

最大值：


````python
print(np.max(a))
````

输出：

```text
8
```

也可以写成数组自己的方法：


````python
print(a.sum())
print(a.min())
print(a.max())
````

输出：

```text
36
1
8
```

## 13. `axis=0`是什么意思

原数组：

```text
[[1 2 3 4]
 [5 6 7 8]]
```

执行：


````python
print(np.sum(a, axis=0))
````

输出：

```text
[ 6  8 10 12]
```

`axis=0` 会把上下方向的行压缩掉，得到每一列的结果：

```text
第1列：1 + 5 = 6
第2列：2 + 6 = 8
第3列：3 + 7 = 10
第4列：4 + 8 = 12
```

所以：

```text
axis=0 → 按列统计
```

结果有4个数，因为原来有4列。

---

## 14. `axis=1`是什么意思


````python
print(np.sum(a, axis=1))
````

输出：

```text
[10 26]
```

分别计算每一行：

```text
第1行：1 + 2 + 3 + 4 = 10
第2行：5 + 6 + 7 + 8 = 26
```

所以：

```text
axis=1 → 按行统计
```

结果有2个数，因为原来有2行。

---

## 15. axis完整示例


````python
a = np.array([
    [1, 2, 3, 4],
    [5, 6, 7, 8]
])

print("所有元素求和：", np.sum(a))
print("每列求和：", np.sum(a, axis=0))
print("每行求和：", np.sum(a, axis=1))

print("每列最小值：", np.min(a, axis=0))
print("每行最大值：", np.max(a, axis=1))
````

输出：

```text
所有元素求和： 36
每列求和： [ 6  8 10 12]
每行求和： [10 26]
每列最小值： [1 2 3 4]
每行最大值： [4 8]
```

最简单的记忆方式：

```text
axis=0：上下计算，得到每列结果
axis=1：左右计算，得到每行结果
```
