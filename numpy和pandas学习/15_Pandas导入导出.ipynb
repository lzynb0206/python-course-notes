## 1. 读取 CSV

CSV 是最常见的表格交换格式。`Path` 能让路径拼接更清楚。仓库已经提供 `student.csv` 作为示例数据。


```python
from pathlib import Path
import pandas as pd

base_dir = Path.cwd()
csv_path = base_dir / "student.csv"
students = pd.read_csv(csv_path)
students
```

## 2. 常用读取参数

- `usecols`：只读取需要的列。
- `dtype`：明确列类型，防止编号被当成数字。
- `na_values`：指定哪些文本代表缺失。
- `parse_dates`：读取时转换日期列。

实际项目应先查看原文件列名，再设置参数。


```python
selected = pd.read_csv(csv_path, usecols=students.columns[:2].tolist())
selected.head()
```

## 3. 导出 CSV

`index=False` 避免把 DataFrame 的行索引额外保存成一列。中文 CSV 常用 `utf-8-sig`，方便某些表格软件正确识别编码。


```python
output_dir = base_dir / "runtime_data"
output_dir.mkdir(exist_ok=True)
output_csv = output_dir / "students_copy.csv"

students.to_csv(output_csv, index=False, encoding="utf-8-sig")
pd.read_csv(output_csv).head()
```

## 4. JSON 适合与程序交换数据

`orient="records"` 会生成“每行一个对象”的结构，适合 API 或前端使用。JSON 能跨语言，但类型信息不如专用表格格式完整。


```python
json_text = students.head(2).to_json(orient="records", force_ascii=False)
json_text
```

## 5. Pickle 的特点与安全提醒

Pickle 能较完整地保留 Python/Pandas 对象，但主要适合可信环境中的临时保存。**绝不要读取来源不可信的 Pickle 文件**，因为反序列化可能执行恶意代码。


```python
pickle_path = output_dir / "students.pkl"
students.to_pickle(pickle_path)
restored = pd.read_pickle(pickle_path)

restored.equals(students)
```

## 常见格式对照

| 格式 | 优点 | 注意点 |
|---|---|---|
| CSV | 通用、可读 | 类型信息有限 |
| JSON | 适合程序交换 | 大表体积可能较大 |
| Excel | 适合人工查看 | 需要额外引擎，速度较慢 |
| Pickle | 保留 Python 对象 | 仅加载可信文件，不跨语言 |
