## 1. 读取CSV

假设 `student.csv` 内容是：

```csv
name,age,score
小明,18,90
小红,19,85
小刚,20,92
```

读取：


````python
import pandas as pd

data = pd.read_csv("student.csv")

print(data)
````

输出：

```text
  name  age  score
0   小明   18     90
1   小红   19     85
2   小刚   20     92
```

---

## 2. `read_csv()`返回什么


````python
print(type(data))
````

输出：

```text
<class 'pandas.DataFrame'>
```

读取文件以后，就可以像普通DataFrame一样处理：


````python
print(data["score"].mean())
````

输出：

```text
89.0
```

筛选：


````python
print(data[data["score"] >= 90])
````

输出：

```text
  name  age  score
0   小明   18     90
2   小刚   20     92
```

---

## 3. 指定某列作为索引


````python
data = pd.read_csv(
    "student.csv",
    index_col="name"
)

print(data)
````

输出：

```text
      age  score
name
小明     18     90
小红     19     85
小刚     20     92
```

现在可以通过姓名标签选择：

```python
print(data.loc["小红"])
```

输出：

```text
age      19
score    85
Name: 小红, dtype: int64
```

## 4. 保存为CSV

```python
data.to_csv(
    "result.csv",
    index=False,
    encoding="utf-8-sig"
)
```

参数：

```text
"result.csv"      → 保存路径
index=False       → 不把DataFrame索引保存成额外一列
encoding=utf-8-sig → Windows Excel打开中文时通常更稳妥
```

如果不写：

```python
index=False
```

默认索引可能也会被写进CSV。

## 6. 保存为Pickle

原视频演示：

```python
data.to_pickle("student.pickle")
```

读取回来：

```python
new_data = pd.read_pickle("student.pickle")

print(new_data)
```

Pickle会保留较多Python和Pandas类型信息，但不适合与其他编程语言交换数据。

安全提醒：

> 不要读取来源不可信的Pickle文件，因为反序列化恶意Pickle可能执行危险代码。

如果只是跨软件交换表格，CSV通常更通用。

## 可运行的安全导出练习

原聊天记录中的 `result.csv`、`student.pickle` 写法保留在上面的讲解中。这里把同样的操作放到 `runtime_data/`，并立即读回检查。


````python
from pathlib import Path

data = pd.read_csv("student.csv")
output_dir = Path("runtime_data")
output_dir.mkdir(exist_ok=True)

csv_output = output_dir / "result.csv"
pickle_output = output_dir / "student.pickle"

data.to_csv(csv_output, index=False, encoding="utf-8-sig")
data.to_pickle(pickle_output)

csv_data = pd.read_csv(csv_output)
pickle_data = pd.read_pickle(pickle_output)
print(csv_data.head())
print("Pickle读取结果与原数据相同：", pickle_data.equals(data))
````
