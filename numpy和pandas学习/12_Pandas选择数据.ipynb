## 1. 准备数据


````python
import pandas as pd
import numpy as np

dates = pd.date_range("20130101", periods=6)

df = pd.DataFrame(
    np.arange(24).reshape((6, 4)),
    index=dates,
    columns=["A", "B", "C", "D"]
)

print(df)
````

输出：

```text
             A   B   C   D
2013-01-01   0   1   2   3
2013-01-02   4   5   6   7
2013-01-03   8   9  10  11
2013-01-04  12  13  14  15
2013-01-05  16  17  18  19
2013-01-06  20  21  22  23
```

---

## 2. 根据列名选择一列


````python
print(df["A"])
````

输出：

```text
2013-01-01     0
2013-01-02     4
2013-01-03     8
2013-01-04    12
2013-01-05    16
2013-01-06    20
Freq: D, Name: A, dtype: int64
```

也可以写：

```python
df.A
```

但更推荐：

```python
df["A"]
```

因为如果列名包含空格、中文特殊符号或与方法重名，点号写法可能不能使用。

---

## 3. 选择多列


````python
print(df[["A", "C"]])
````

输出：

```text
             A   C
2013-01-01   0   2
2013-01-02   4   6
2013-01-03   8  10
2013-01-04  12  14
2013-01-05  16  18
2013-01-06  20  22
```

外层中括号是 DataFrame 索引语法；内层列表：

```python
["A", "C"]
```

表示需要两列。

---

## 4. 根据位置切片行


````python
print(df[0:3])
````

输出：

```text
            A  B   C   D
2013-01-01  0  1   2   3
2013-01-02  4  5   6   7
2013-01-03  8  9  10  11
```

和普通 Python 切片一样：

```text
包含位置0
不包含位置3
```

---

## 5. 根据日期标签切片


````python
print(df.loc["20130102":"20130104"])
````

输出：

```text
             A   B   C   D
2013-01-02   4   5   6   7
2013-01-03   8   9  10  11
2013-01-04  12  13  14  15
```

这里需要特别注意：

```text
普通位置切片通常不包含结束位置
loc标签切片通常包含结束标签
```

所以 `"20130104"` 这一行也被选出来。Pandas官方文档也明确说明，`.loc` 标签切片的起点和终点通常都包含在结果中。[Pandas索引文档](https://pandas.pydata.org/docs/user_guide/indexing.html)

---

## 6. `loc`根据标签选择

选择某一行：


````python
print(df.loc["20130102"])
````

输出：

```text
A    4
B    5
C    6
D    7
Name: 2013-01-02 00:00:00, dtype: int64
```

选择所有行的A、B列：


````python
print(df.loc[:, ["A", "B"]])
````

输出：

```text
             A   B
2013-01-01   0   1
2013-01-02   4   5
2013-01-03   8   9
2013-01-04  12  13
2013-01-05  16  17
2013-01-06  20  21
```

这里：

```text
:          → 所有行
["A","B"]  → A列和B列
```

选择一行中的两列：


````python
print(df.loc["20130102", ["A", "B"]])
````

输出：

```text
A    4
B    5
Name: 2013-01-02 00:00:00, dtype: int64
```

---

## 7. `iloc`根据位置选择

取第4行、第2列：


````python
print(df.iloc[3, 1])
````

输出：

```text
13
```

位置从0开始：

```text
行位置3 → 第4行
列位置1 → 第2列B
```

对应：

```text
2013-01-04的B列 → 13
```

---

## 8. `iloc`连续切片


````python
print(df.iloc[3:5, 1:3])
````

输出：

```text
             B   C
2013-01-04  13  14
2013-01-05  17  18
```

拆开：

```text
3:5 → 行位置3、4
1:3 → 列位置1、2
```

位置切片不包含结束位置。

---

## 9. `iloc`选择不连续的行


````python
print(df.iloc[[1, 3, 5], 1:3])
````

输出：

```text
             B   C
2013-01-02   5   6
2013-01-04  13  14
2013-01-06  21  22
```

```python
[1, 3, 5]
```

表示只选择行位置1、3、5。

---

## 10. 视频中的 `.ix` 已经不能用了

原视频中有：

```python
df.ix[:3, ["A", "C"]]
```

`.ix` 过去用于混合标签与位置，但已经从 Pandas 中删除。

新版中应该明确选择：


````python
# 按位置选前三行，再选A、C列
print(df.iloc[:3][["A", "C"]])
````

更清晰的写法：


````python
print(df.loc[df.index[:3], ["A", "C"]])
````

输出：

```text
            A   C
2013-01-01  0   2
2013-01-02  4   6
2013-01-03  8  10
```

以后只记：

```text
loc  → 标签
iloc → 位置
```

不要再使用 `.ix`。

## 11. 条件筛选

筛选A列大于8的行：


````python
condition = df["A"] > 8

print(condition)
````

输出：

```text
2013-01-01    False
2013-01-02    False
2013-01-03    False
2013-01-04     True
2013-01-05     True
2013-01-06     True
Freq: D, Name: A, dtype: bool
```

用条件筛选整个DataFrame：


````python
print(df[df["A"] > 8])
````

输出：

```text
             A   B   C   D
2013-01-04  12  13  14  15
2013-01-05  16  17  18  19
2013-01-06  20  21  22  23
```

执行过程：

```text
先判断每一行的A是否大于8
          ↓
得到True和False
          ↓
只保留True对应的整行
```

多个条件：


````python
print(df[(df["A"] > 8) & (df["D"] < 20)])
````

输出：

```text
             A   B   C   D
2013-01-04  12  13  14  15
2013-01-05  16  17  18  19
```

Pandas多个条件需要：

```text
& → 并且
| → 或者
~ → 取反
```

并且每个条件应加括号。
