## 1. 准备带日期索引的数据

下面用日期作为行标签。观察标签之后再选择，能避免把“第几行”和“标签是什么”混在一起。


```python
import numpy as np
import pandas as pd

dates = pd.date_range("2026-01-01", periods=6)
df = pd.DataFrame(
    np.arange(24).reshape(6, 4),
    index=dates,
    columns=["A", "B", "C", "D"],
)
df
```

## 2. 选择一列或多列

`df["A"]` 返回 `Series`；`df[["A", "C"]]` 返回 `DataFrame`。双层方括号不是特殊语法，外层是选择，内层是列名列表。


```python
one_column = df["A"]
two_columns = df[["A", "C"]]
one_column, two_columns
```

## 3. `loc` 按标签选择

`loc[行标签, 列标签]`。标签切片通常包含结束标签，这和普通 Python 切片不同。


```python
by_label = df.loc["2026-01-02":"2026-01-04", ["A", "C"]]
by_label
```

## 4. `iloc` 按整数位置选择

`iloc[行位置, 列位置]` 使用从 0 开始的位置，切片仍然左闭右开。它不关心行列标签长什么样。


```python
by_position = df.iloc[1:4, [0, 2]]
by_position
```

## 5. 条件筛选

多个条件要分别加括号，并使用 `&`（且）、`|`（或）、`~`（非），不能直接使用 Python 的 `and/or/not`。


```python
filtered = df.loc[(df["A"] >= 8) & (df["D"] < 20), ["A", "D"]]
filtered
```

## 6. 单个值：`at` 与 `iat`

只读取一个单元格时，`at` 按标签，`iat` 按位置，语义比通用选择器更明确。


```python
label_value = df.at[dates[0], "B"]
position_value = df.iat[0, 1]
label_value, position_value
```

## 旧教程中的 `ix`

早期课程可能使用 `df.ix[...]`，但它混合标签和位置，含义容易歧义，已从 Pandas 移除。现代代码应明确使用 `loc` 或 `iloc`。


## 本节小结

记住一句话：`loc` 看标签，`iloc` 看位置。条件筛选通常和 `loc` 组合使用，既能筛行，也能同时指定需要的列。
