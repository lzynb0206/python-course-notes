```python
# Basic, unlabelled Series creation:
import pandas as pd
sr1 = pd.Series([3, 4, 5])
sr1
# Basic, labelled Series creation:
sr2 = pd.Series([3, 4, 5], index=['one', 'two', 'three'])
sr2
# Labelled Series from dictionary data:
dc = {'a': 1, 'b': 2, 'c': 3, 'd': 4}
sr = pd.Series(dc)
sr
```

```python
# Basic manual DataFrame creation, automatic index values
df1 = pd.DataFrame(data=([[1, 2, 'r'],
 [3, 4, 's'],
 [5, 6, 't'],
 [7, 8, 'u'],
 [9, 10, 'v']]),
 columns=['column_1', 'column_2', 'column_3'])
df1
# DataFrame creation from a dictionary of lists:
d = {'dict_column_1': [1, 3, 5, 7, 9],
 'dict_column_2': [2, 4, 6, 8, 10],
 'dict_column_3': ['r','s','t','u','v']}
df2 = pd.DataFrame(d)
df2
# Inspect data types
df2.dtypes
```
