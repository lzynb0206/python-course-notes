## 1. 一维索引与切片

索引从 0 开始，负数从末尾开始。切片遵循“左闭右开”，`[1:4]` 包含位置 1、2、3，不包含位置 4。


```python
import numpy as np

a = np.arange(10, 60, 10)
a[0], a[-1], a[1:4], a[::-1]
```

## 2. 二维索引

推荐写成 `array[行, 列]`。冒号表示该轴全部保留，因此 `matrix[:, 1]` 是第 2 列，`matrix[1, :]` 是第 2 行。


```python
matrix = np.arange(1, 13).reshape(3, 4)

second_row = matrix[1, :]
second_column = matrix[:, 1]
block = matrix[:2, 1:3]
matrix, second_row, second_column, block
```

## 3. 布尔索引

比较表达式会生成和原数组同形状的布尔数组。把它放进方括号，就能取出所有满足条件的元素。


```python
mask = matrix % 2 == 0
even_numbers = matrix[mask]

mask, even_numbers
```

## 4. 花式索引

用整数列表可以按指定顺序选取多行。花式索引会创建副本，这一点和基础切片不同。


```python
selected_rows = matrix[[2, 0]]
selected_elements = matrix[[0, 2], [1, 3]]

selected_rows, selected_elements
```

## 5. 遍历行、列和所有元素

直接遍历二维数组时，每次得到一行；遍历 `matrix.T` 时每次得到一列；`flat` 是按元素遍历的迭代器。


```python
rows = [row.tolist() for row in matrix]
columns = [column.tolist() for column in matrix.T]
elements = list(matrix.flat)

rows, columns, elements
```

## 6. `ravel()` 与 `flatten()`

两者都能把数组变成一维。`ravel()` 尽量返回共享数据的视图；`flatten()` 总是返回独立副本。是否共享内存可用 `np.shares_memory()` 检查。


```python
raveled = matrix.ravel()
flattened = matrix.flatten()

np.shares_memory(matrix, raveled), np.shares_memory(matrix, flattened)
```

## 本节小结

二维选择先写行、再写列。基础切片常与原数组共享数据，布尔索引和花式索引通常返回副本；修改前要清楚自己拿到的是什么。
