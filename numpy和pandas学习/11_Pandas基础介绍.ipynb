## 1. 导入Pandas和NumPy


````python
import pandas as pd
import numpy as np
````

- `pd` 是 Pandas 约定俗成的简称
- `np` 是 NumPy 约定俗成的简称

---

## 2. Series是什么


````python
s = pd.Series([1, 3, 6, np.nan, 44, 1])

print(s)
````

输出：

```text
0     1.0
1     3.0
2     6.0
3     NaN
4    44.0
5     1.0
dtype: float64
```

Series可以理解成两部分：

```text
左边：索引index
右边：数据value
```

也就是：

```text
索引    数据
 0      1.0
 1      3.0
 2      6.0
 3      NaN
 4     44.0
 5      1.0
```

没有指定索引时，Pandas自动生成：

```text
0、1、2、3、4、5
```

---

## 3. 为什么整数变成了小数

原始数据是：

```python
[1, 3, 6, np.nan, 44, 1]
```

由于里面有：

```python
np.nan
```

传统 NumPy 整数类型不能表示 `NaN`，所以整组数据被转换为浮点数：

```text
1  → 1.0
3  → 3.0
6  → 6.0
44 → 44.0
```

因此显示：

```text
dtype: float64
```

## 4. 自己指定Series索引


````python
scores = pd.Series(
    [90, 85, 92],
    index=["小明", "小红", "小刚"]
)

print(scores)
````

输出：

```text
小明    90
小红    85
小刚    92
dtype: int64
```

通过标签取值：


````python
print(scores["小红"])
````

输出：

```text
85
```

修改：


````python
scores["小红"] = 100

print(scores)
````

输出：

```text
小明     90
小红    100
小刚     92
dtype: int64
```

可以把 Series 理解为：

```python
{
    "小明": 90,
    "小红": 100,
    "小刚": 92
}
```

但 Series 比字典更适合批量数据计算。

---

## 5. Series批量运算


````python
scores = pd.Series(
    [90, 85, 92],
    index=["小明", "小红", "小刚"]
)

print(scores + 5)
````

输出：

```text
小明    95
小红    90
小刚    97
dtype: int64
```

每个元素都加5。

条件判断：


````python
print(scores >= 90)
````

输出：

```text
小明     True
小红    False
小刚     True
dtype: bool
```

条件筛选：


````python
print(scores[scores >= 90])
````

输出：

```text
小明    90
小刚    92
dtype: int64
```

## 6. DataFrame是什么

DataFrame是带有行名和列名的二维表格。


````python
df = pd.DataFrame(
    [
        [90, 85],
        [78, 92],
        [88, 76]
    ],
    index=["小明", "小红", "小刚"],
    columns=["语文", "数学"]
)

print(df)
````

输出：

```text
    语文  数学
小明  90  85
小红  78  92
小刚  88  76
```

结构：

```text
          列名columns
           语文  数学
行名 小明   90   85
index 小红  78   92
     小刚   88   76
```

## 7. 使用日期作为索引

原视频使用：


````python
dates = pd.date_range("20160101", periods=6)

print(dates)
````

输出类似：

```text
DatetimeIndex(['2016-01-01', '2016-01-02', '2016-01-03',
               '2016-01-04', '2016-01-05', '2016-01-06'],
              dtype='datetime64[us]', freq='D')
```

`periods=6` 表示从指定日期开始创建6个日期。

创建DataFrame：


````python
df = pd.DataFrame(
    np.arange(24).reshape(6, 4),
    index=dates,
    columns=["A", "B", "C", "D"]
)

print(df)
````

输出：

```text
             A   B   C   D
2016-01-01   0   1   2   3
2016-01-02   4   5   6   7
2016-01-03   8   9  10  11
2016-01-04  12  13  14  15
2016-01-05  16  17  18  19
2016-01-06  20  21  22  23
```

---

## 8. 选择一列


````python
print(df["B"])
````

输出：

```text
2016-01-01     1
2016-01-02     5
2016-01-03     9
2016-01-04    13
2016-01-05    17
2016-01-06    21
Freq: D, Name: B, dtype: int64
```

一列 DataFrame 取出来后，通常变成 Series：


````python
print(type(df["B"]))
````

输出：

```text
<class 'pandas.Series'>
```

如果想保留二维 DataFrame：


````python
print(type(df[["B"]]))
````

输出：

```text
<class 'pandas.DataFrame'>
```

注意中括号数量：

```python
df["B"]    # Series
df[["B"]]  # DataFrame
```

## 9. 使用字典创建DataFrame


````python
df2 = pd.DataFrame({
    "姓名": ["小明", "小红", "小刚"],
    "年龄": [18, 19, 20],
    "及格": [True, True, False]
})

print(df2)
````

输出：

```text

0  小明  18   True
1  小红  19   True
2  小刚  20  False
```

字典中的：

```text
键   → 列名
列表 → 这一列的数据
```

## 11. `describe()`数据统计


````python
scores = pd.DataFrame({
    "语文": [90, 80, 70],
    "数学": [60, 80, 100]
})

print(scores.describe())
````

输出：

```text
         语文     数学
count   3.0    3.0
mean   80.0   80.0
std    10.0   20.0
min    70.0   60.0
25%    75.0   70.0
50%    80.0   80.0
75%    85.0   90.0
max    90.0  100.0
```

主要字段：

| 字段 | 含义 |
|---|---|
| `count` | 非空数据数量 |
| `mean` | 平均值 |
| `std` | 标准差 |
| `min` | 最小值 |
| `25%` | 第一四分位数 |
| `50%` | 中位数 |
| `75%` | 第三四分位数 |
| `max` | 最大值 |

---

## 12. 转置和排序

转置：


````python
print(scores.T)
````

输出：

```text
      0    1    2
语文   90   80   70
数学   60   80  100
```

按照列名排序：


````python
print(scores.sort_index(axis=1, ascending=False))
````

如果列名是“语文、数学”，会按照列名降序重新排列。

按照数据值排序：


````python
print(scores.sort_values(by="语文"))
````

输出：

```text
   语文   数学
2  70  100
1  80   80
0  90   60
```

`sort_values(by="语文")` 表示按照语文成绩从小到大排列整行。
