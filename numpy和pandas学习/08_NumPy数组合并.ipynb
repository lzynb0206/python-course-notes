## 1. 准备两个一维数组

合并前要先看形状。一维数组只有一个轴，因此“竖着拼”时结果会出现新的二维结构。


```python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
a.shape, b.shape
```

## 2. `vstack()` 与 `hstack()`

`vstack` 按垂直方向堆叠，两个一维数组变成两行；`hstack` 水平连接，结果仍是一维。


```python
vertical = np.vstack((a, b))
horizontal = np.hstack((a, b))

vertical, vertical.shape, horizontal, horizontal.shape
```

## 3. 用 `newaxis` 明确行和列

`np.newaxis`（也可写 `None`）会在指定位置增加长度为 1 的轴。`a[:, None]` 是列向量，`a[None, :]` 是行向量。


```python
column_a = a[:, np.newaxis]
row_a = a[np.newaxis, :]

column_a, column_a.shape, row_a, row_a.shape
```

## 4. `concatenate()` 沿已有轴连接

`axis=0` 增加行，`axis=1` 增加列。除拼接轴外，其他维度必须相同。


```python
left = np.array([[1, 2], [3, 4]])
right = np.array([[5, 6], [7, 8]])

by_rows = np.concatenate((left, right), axis=0)
by_columns = np.concatenate((left, right), axis=1)
by_rows, by_columns
```

## 5. `stack()` 创建新轴

`concatenate` 沿已有轴变长，`stack` 则增加一个新轴。原数组形状相同才可 `stack`。


```python
stacked_axis0 = np.stack((left, right), axis=0)
stacked_axis2 = np.stack((left, right), axis=2)

stacked_axis0.shape, stacked_axis2.shape
```

## 合并方法对照

| 方法 | 结果特点 |
|---|---|
| `hstack` | 水平组合 |
| `vstack` | 垂直组合 |
| `concatenate` | 沿已有轴连接 |
| `stack` | 新建一个轴再堆叠 |


## 本节小结

合并错误几乎都可以通过检查 `shape` 解释。先决定是让某个已有轴变长，还是创建新轴，再选择 `concatenate` 或 `stack`。
