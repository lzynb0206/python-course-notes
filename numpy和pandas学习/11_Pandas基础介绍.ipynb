## 1. `Series`：带标签的一维数据

`Series` 同时保存数值和索引。没有指定索引时默认使用 0、1、2……；指定标签后可以用标签访问。


```python
import numpy as np
import pandas as pd

scores = pd.Series([88, 92, 79], index=["小林", "小周", "小陈"], name="成绩")
scores
```

## 2. `DataFrame`：带行列标签的二维表

字典的键会成为列名，每个列表成为一列。表格可以同时包含数字、字符串、布尔值等不同类型。


```python
students = pd.DataFrame(
    {
        "姓名": ["小林", "小周", "小陈"],
        "成绩": [88, 92, 79],
        "已通过": [True, True, False],
    },
    index=["S001", "S002", "S003"],
)
students
```

## 3. 行索引、列名与数据类型

- `index` 是行标签。
- `columns` 是列标签。
- `dtypes` 展示每一列的类型。
- `shape` 返回 `(行数, 列数)`。


```python
students.index, students.columns, students.dtypes, students.shape
```

## 4. 快速了解一张陌生表

`head()` 和 `tail()` 查看首尾记录，`describe()` 汇总数值列，`info()` 展示列类型、非空数量和内存信息。


```python
students.head(2)
```

`describe(include="all")` 也会统计文本和布尔列。不同类型的列会显示不同统计项。


```python
students.describe(include="all")
```

## 5. 转置和排序

`.T` 交换行列；`sort_values()` 按某列的值排序；`sort_index()` 按标签排序。排序默认返回新表，不会直接修改原表。


```python
sorted_students = students.sort_values("成绩", ascending=False)
sorted_students
```
