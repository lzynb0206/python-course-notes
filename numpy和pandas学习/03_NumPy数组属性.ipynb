## 1. 创建二维数组

下面的数据有 2 行、3 列。二维数组有两个轴：`axis=0` 对应行方向，`axis=1` 对应列方向。


```python
import numpy as np

a = np.array([[1, 2, 3], [4, 5, 6]])
a
```

## 2. 四个最常用属性

- `ndim`：轴的数量，也就是数组是几维。
- `shape`：每个轴的长度，这里是 `(2, 3)`。
- `size`：元素总数，等于各维长度之积。
- `dtype`：每个元素的数据类型。


```python
{
    "维度 ndim": a.ndim,
    "形状 shape": a.shape,
    "元素数 size": a.size,
    "类型 dtype": str(a.dtype),
    "单元素字节数 itemsize": a.itemsize,
}
```

## 3. 一维、二维和三维

方括号的嵌套层数有助于直观判断维度，但最终应以 `ndim` 和 `shape` 为准。三维数组可以理解为“多张二维表叠在一起”。


```python
one_d = np.array([1, 2, 3])
two_d = np.array([[1, 2, 3], [4, 5, 6]])
three_d = np.arange(24).reshape(2, 3, 4)

[(x.ndim, x.shape) for x in (one_d, two_d, three_d)]
```

## 4. `reshape()` 只改变形状

只要元素总数不变，就可以重新排列形状。`-1` 表示让 NumPy 自动推断这一维的长度。


```python
original = np.arange(12)
matrix = original.reshape(3, 4)
inferred = original.reshape(2, -1)

matrix, inferred
```

## 易错点：`(3,)` 不是 `(1, 3)`

一维数组没有“行向量/列向量”的方向。需要矩阵形状时，应显式增加维度。


```python
vector = np.array([1, 2, 3])
row = vector.reshape(1, 3)
column = vector.reshape(3, 1)

vector.shape, row.shape, column.shape
```

## 本节小结

`shape` 是后续索引、广播、拼接和矩阵运算的核心。看到报错时，第一件事通常是检查参与运算的数组形状。
