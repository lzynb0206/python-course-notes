## 1. 准备两张结构相同的表

`concat` 的本质是沿某个轴把对象排列在一起。按行拼接时，通常要求列结构相近。


```python
import pandas as pd

first = pd.DataFrame({"姓名": ["小林", "小周"], "成绩": [88, 92]})
second = pd.DataFrame({"姓名": ["小陈", "小吴"], "成绩": [79, 85]})
first, second
```

## 2. `axis=0` 按行拼接

默认沿行方向拼接。原表索引可能重复，`ignore_index=True` 会重新生成连续索引。


```python
by_rows = pd.concat([first, second], ignore_index=True)
by_rows
```

## 3. 列不完全一致时

默认 `join="outer"` 保留所有列，缺少的位置填 `NaN`；`join="inner"` 只保留各表共有的列。


```python
extra = pd.DataFrame({"姓名": ["小郑"], "城市": ["深圳"]})

outer = pd.concat([first, extra], join="outer", ignore_index=True)
inner = pd.concat([first, extra], join="inner", ignore_index=True)
outer, inner
```

## 4. `axis=1` 按列拼接

按列拼接时，Pandas 会根据行索引对齐。两个表的索引不一致时，结果会出现缺失值。


```python
scores = pd.DataFrame({"成绩": [88, 92]}, index=["S001", "S002"])
cities = pd.DataFrame({"城市": ["北京", "广州"]}, index=["S001", "S003"])

side_by_side = pd.concat([scores, cities], axis=1)
side_by_side
```

## 5. 先 `reindex()` 再按列拼接

旧教程可能出现已经移除的 `join_axes` 参数。现代写法是先把参与拼接的表 `reindex` 到目标索引，再调用 `concat`。


```python
aligned_cities = cities.reindex(scores.index)
aligned = pd.concat([scores, aligned_cities], axis=1)
aligned
```

## 6. 用 `keys` 保留来源

`keys` 会创建分层索引，适合在合并后仍要区分数据来源的场景。


```python
with_source = pd.concat([first, second], keys=["一班", "二班"])
with_source
```

## 旧写法提醒

早期代码常用 `DataFrame.append()` 追加行，该方法已移除。统一改用 `pd.concat([df1, df2], ignore_index=True)`。


## 本节小结

`concat` 负责沿轴拼接，关键是轴方向和索引对齐。按行拼常用 `ignore_index=True`；按列拼之前先检查行索引。
