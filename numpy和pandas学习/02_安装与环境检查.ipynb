```bash
pip3 install numpy
```

```bash
python3 -m pip install numpy pandas
```

NumPy 官方目前仍支持通过 `pip` 或 `conda` 安装，并建议使用虚拟环境管理项目依赖。[NumPy官方安装说明](https://numpy.org/install/)

## 1. macOS安装

打开终端，输入：

```bash
python3 -m pip install numpy pandas
```

安装完成后验证：

```bash
python3 -c "import numpy; import pandas; print(numpy.__version__); print(pandas.__version__)"
```

输出类似：

```text
2.2.6
2.3.2
```

你的版本号可能不同，只要没有报错即可。

## 4. `import numpy as np`是什么意思


````python
import numpy as np
````

分开理解：

```text
import numpy：导入numpy模块
as np：把numpy临时简称为np
```

因此：

```python
np.array([1, 2, 3])
```

相当于：

```python
numpy.array([1, 2, 3])
```

`np` 是行业中约定俗成的写法，不是强制关键字。

理论上可以写：


````python
import numpy as abc

data = abc.array([1, 2, 3])
````

也能运行，但别人看起来会很不习惯，所以推荐使用：


````python
import numpy as np
````

Pandas通常写：


````python
import pandas as pd
````

## 5. 常见安装问题

### 问题一：找不到pip

```text
pip: command not found
```

尝试：

```bash
python3 -m pip --version
```

或者 Windows：

```bash
python -m pip --version
```

---

### 问题二：代码里找不到NumPy

```text
ModuleNotFoundError: No module named 'numpy'
```

常见原因是：

> 安装 NumPy 的 Python，和运行代码的 Python，不是同一个解释器。

例如安装时使用：

```bash
python3.12 -m pip install numpy
```

但运行时使用了另外一个 Python。

在代码中查看当前解释器：


````python
import sys

print(sys.executable)
````

输出类似：

```text
/opt/homebrew/bin/python3
```

应当用这个解释器安装：

```bash
/opt/homebrew/bin/python3 -m pip install numpy pandas
```

---

### 问题三：不要照搬旧视频里的 `sudo pip`

原教程网页中有旧式的：

```bash
sudo pip install numpy
```

现在一般不建议直接给系统 Python 使用 `sudo pip`，因为可能破坏系统环境。更推荐：

- 使用虚拟环境
- 使用 Conda
- 使用用户自己的 Python 环境

初学时，如果普通安装能成功，不要加 `sudo`。
