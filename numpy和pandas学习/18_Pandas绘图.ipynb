## 1. 准备绘图数据

Pandas 的 `.plot()` 是 Matplotlib 的便捷封装。先准备带日期索引的 DataFrame，并统一设置图形样式。


```python
from pathlib import Path
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

dates = pd.date_range("2026-01-01", periods=6)
sales = pd.DataFrame(
    {
        "线上": [120, 135, 128, 160, 172, 180],
        "线下": [90, 88, 96, 105, 110, 108],
    },
    index=dates,
)
sales
```

## 2. 折线图：观察趋势

折线图适合连续时间上的变化。`ax` 是 Matplotlib 的坐标区域对象，可以继续设置标题、轴标签和网格。


```python
ax = sales.plot(figsize=(8, 4), marker="o", title="Sales Trend")
ax.set_xlabel("Date")
ax.set_ylabel("Sales")
ax.grid(alpha=0.25)
plt.tight_layout()
plt.show()
```

## 3. 柱状图：比较类别

柱状图适合比较离散类别。这里对各渠道求总和，再绘制一张简单柱状图。


```python
totals = sales.sum()
ax = totals.plot(kind="bar", color=["#4C78A8", "#F58518"], title="Total Sales by Channel")
ax.set_xlabel("Channel")
ax.set_ylabel("Total")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 4. 散点图：观察两个变量的关系

散点图不强调时间顺序，而是观察两列数值是否一起变化、是否存在异常点。


```python
ax = sales.plot.scatter(x="线上", y="线下", s=70, color="#54A24B", title="Online vs Offline")
ax.set_xlabel("Online")
ax.set_ylabel("Offline")
ax.grid(alpha=0.25)
plt.tight_layout()
plt.show()
```

## 5. 一个画布放多张子图

需要并排比较时，可以先用 `plt.subplots()` 创建多个坐标区域，再通过 `ax=` 指定每张图画在哪里。


```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
sales.plot(ax=axes[0], marker="o", title="Trend")
totals.plot(kind="bar", ax=axes[1], color=["#4C78A8", "#F58518"], title="Totals")
axes[1].tick_params(axis="x", rotation=0)
fig.tight_layout()
plt.show()
```

## 6. 保存图片

在 `show()` 之前调用 `savefig()`，并用 `bbox_inches="tight"` 减少标题或标签被裁掉的情况。生成文件放进已忽略的 `runtime_data/`。


```python
output_dir = Path.cwd() / "runtime_data"
output_dir.mkdir(exist_ok=True)
image_path = output_dir / "sales_trend.png"

ax = sales.plot(figsize=(8, 4), marker="o", title="Sales Trend")
ax.set_xlabel("Date")
ax.set_ylabel("Sales")
plt.tight_layout()
plt.savefig(image_path, dpi=150, bbox_inches="tight")
plt.close()

image_path
```

## 图形选择速查

| 问题 | 常用图形 |
|---|---|
| 随时间怎样变化 | 折线图 |
| 不同类别谁高谁低 | 柱状图 |
| 两个数值变量是否相关 | 散点图 |
| 数值分布是否偏斜 | 直方图 / 箱线图 |

Pandas 绘图适合快速探索。需要复杂排版、精细标注时，可以继续使用返回的 `ax` 调用 Matplotlib 接口。
