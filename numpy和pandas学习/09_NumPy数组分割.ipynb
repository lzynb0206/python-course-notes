## 1. 准备二维数组

下面数组有 3 行、4 列。分割时的 `axis` 表示沿哪个轴切开。


```python
import numpy as np

a = np.arange(1, 13).reshape(3, 4)
a
```

## 2. `split()` 严格等分

`np.split(a, 2, axis=1)` 把 4 列平均分成 2 份，每份 2 列。若不能整除，`split()` 会直接报错。


```python
left, right = np.split(a, 2, axis=1)
left, right
```

## 3. 不能整除时用 `array_split()`

3 行无法严格平均分成 2 份。`array_split()` 允许不等分，会让前面的分块尽量多一个元素。


```python
uneven = np.array_split(a, 2, axis=0)
[part.shape for part in uneven], uneven
```

## 4. 按边界位置切分

第二个参数也可以是索引列表。`[1, 3]` 表示在第 1 列和第 3 列之前切开，得到列区间 `[:1]`、`[1:3]` 和 `[3:]`。


```python
parts = np.split(a, [1, 3], axis=1)
parts
```

## 5. `vsplit()` 与 `hsplit()`

`vsplit(a, n)` 等价于沿 `axis=0` 分割；`hsplit(a, n)` 等价于沿 `axis=1` 分割。它们更直观，但仍然要求能整除。


```python
row_parts = np.vsplit(a, 3)
column_parts = np.hsplit(a, 2)

[part.shape for part in row_parts], [part.shape for part in column_parts]
```

## 6. 用 `try/except` 观察典型错误

学习时可以主动捕获一次错误，看看提示。这里不会中断 Notebook。


```python
try:
    np.split(a, 2, axis=0)
except ValueError as error:
    error_message = str(error)

error_message
```

## 本节小结

必须等分时用 `split`，允许不等分时用 `array_split`；行分割看 `axis=0`，列分割看 `axis=1`。
