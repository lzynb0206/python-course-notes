## 1. 为什么使用虚拟环境

不同项目可能依赖不同版本的库。虚拟环境把本项目的依赖隔离起来，避免升级一个项目时影响另一个项目。

在仓库根目录执行：

```bash
python3 -m venv .venv
.venv/bin/python -m pip install -r "numpy和pandas学习/requirements.txt"
```

Windows 的激活和运行路径通常是 `.venv\Scripts\python.exe`。


## 2. 查看当前解释器和版本

Notebook 使用的解释器不一定与终端相同。先查看路径和版本，排查“终端能导入、Notebook 却不能导入”的问题。


````python
import sys
import numpy as np
import pandas as pd
import matplotlib

{
    "Python 路径": sys.executable,
    "Python 版本": sys.version.split()[0],
    "NumPy": np.__version__,
    "Pandas": pd.__version__,
    "Matplotlib": matplotlib.__version__,
}
````

## 3. 最小可运行检查

版本号能显示不代表所有基本功能都正常。这里分别创建数组和表格，并用断言检查结果。断言没有报错，就说明当前环境可以继续学习。


````python
numbers = np.array([1, 2, 3])
table = pd.DataFrame({"数字": numbers, "平方": numbers ** 2})

assert numbers.sum() == 6
assert table["平方"].tolist() == [1, 4, 9]
table
````

## 常见问题

- `ModuleNotFoundError`：当前解释器没有安装该库，先确认上面显示的 Python 路径。
- Notebook 内核错误：在编辑器中把内核切换到本项目的 `.venv`。
- 不建议把自己的虚拟环境提交到 Git；仓库只保存 `requirements.txt`。


## 本节小结

先确认解释器路径，再确认库版本，最后运行最小示例。排查环境问题时，这个顺序最省时间。
