## 1. 逐元素运算

相同形状的数组做 `+ - * / **` 时，对应位置分别计算。结果仍然保持原来的形状。


```python
import numpy as np

a = np.array([10, 20, 30, 40])
b = np.array([1, 2, 3, 4])

a + b, a - b, a * b, a / b, b ** 2
```

## 2. 标量与数组

一个数字和数组运算时，这个数字会“广播”到数组的每个位置。温度转换、统一折扣等都可以直接写成公式。


```python
prices = np.array([50.0, 80.0, 120.0])
discounted = prices * 0.9
discounted
```

## 3. 通用函数（ufunc）

`sin`、`sqrt`、`exp` 等函数会逐元素处理数组。需要注意定义域，例如负数开平方会产生无效值。


```python
angles = np.array([0, np.pi / 2, np.pi])
np.sin(angles), np.sqrt(np.array([1, 4, 9]))
```

## 4. `*` 与 `@` 的区别

`*` 是对应元素相乘；`@` 是矩阵乘法。矩阵乘法要求左侧的列数等于右侧的行数。


```python
left = np.array([[1, 2], [3, 4]])
right = np.array([[5, 6], [7, 8]])

elementwise = left * right
matrix_product = left @ right
elementwise, matrix_product
```

## 5. 广播：行向量加到每一行

NumPy 从形状末尾开始比较：两个维度相等，或其中一个为 `1`，才可以广播。`(2, 3)` 与 `(3,)` 可以运算，因为最后一维都是 3。


```python
matrix = np.array([[1, 2, 3], [4, 5, 6]])
row_offset = np.array([10, 20, 30])

matrix + row_offset
```

## 6. 列向量广播

要给每一行加不同的值，需要把 `(2,)` 改成 `(2, 1)`。`[:, None]` 是增加维度的常用写法。


```python
row_offset = np.array([100, 200])[:, None]
matrix + row_offset
```

## 本节小结

先区分逐元素运算与矩阵运算，再检查形状是否满足广播规则。复杂计算出错时，把每个数组的 `shape` 写出来通常就能定位问题。
