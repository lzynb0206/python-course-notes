## 1. 准备示例数组

这节会反复使用同一个 `3 × 4` 数组。先观察形状和数值分布。


```python
import numpy as np

a = np.array([[1, 8, 3, 4], [9, 2, 7, 6], [5, 0, 11, 10]])
a
```

## 2. 极值与极值位置

`min/max` 返回数值，`argmin/argmax` 返回把数组展平后的索引。`unravel_index()` 可以把平面索引还原成行列坐标。


```python
flat_index = a.argmax()
position = np.unravel_index(flat_index, a.shape)

{"最小值": a.min(), "最大值": a.max(), "最大值平面索引": flat_index, "最大值坐标": position}
```

## 3. 平均数、加权平均数和中位数

普通平均数让每个数据权重相同；`np.average()` 可以传入权重；中位数按大小排序后取中间位置，对极端值更不敏感。

注意：NumPy 数组没有通用的 `a.median()` 方法，应写 `np.median(a)`。


```python
scores = np.array([60, 70, 80, 100])
weights = np.array([1, 1, 2, 6])

{
    "平均数": scores.mean(),
    "加权平均数": np.average(scores, weights=weights),
    "中位数": np.median(scores),
}
```

## 4. `axis` 决定压缩哪个轴

对二维数组：`axis=0` 压缩行轴，得到每一列的结果；`axis=1` 压缩列轴，得到每一行的结果。可以理解为“哪个轴消失了”。


```python
column_sums = a.sum(axis=0)
row_sums = a.sum(axis=1)

column_sums, row_sums
```

## 5. 累计和与相邻差

`cumsum()` 把前面的数不断累加；`diff()` 计算相邻元素之差，所以结果会少一个元素。指定 `axis` 后可以按行或按列处理。


```python
sequence = np.array([3, 5, 10, 18])
cumulative = np.cumsum(sequence)
differences = np.diff(sequence)

cumulative, differences
```

## 6. 找位置、排序、转置与裁剪

- `nonzero()` 返回非零元素在每个轴上的索引。
- `sort()` 默认对最后一个轴排序。
- `.T` 交换二维数组的行列。
- `clip(low, high)` 把过小或过大的值限制到边界。


```python
nonzero_positions = np.nonzero(a)
sorted_rows = np.sort(a, axis=1)
transposed = a.T
clipped = np.clip(a, 3, 8)

nonzero_positions, sorted_rows, transposed, clipped
```

## 本节小结

聚合函数最关键的参数是 `axis`。`argmax` 得到的是位置而不是最大值；二维坐标可以用 `unravel_index` 还原。
